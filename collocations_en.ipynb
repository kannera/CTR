```python
import json, os, re, sys, codecs, time
import matplotlib.pyplot as plt
import pandas, numpy
from scipy.stats import binom
import wget
from eng_corpus import *
```

```python
#set the corpus here. "crop" can be used to concentrate on certain subsections of the data based on the metadata categories. 
#See the data documentation document for description on which categories are available for each dataset
corpus = "coha"
crop = ""
```

```python
#this will take some time, maybe around two minutes or so. Full frequency list for the whole corpus is being built
data, frequencies = build_data_for_collocations(corpus, crop)
```

```python
#set here the studied word and the window size (in words on either side)
lemma = "economy"
rang=5
```

```python
#this will build the collocations
collocations = build_collocations(data, frequencies, lemma, rang, ["pmi", "llr"])
```

```python
#set the test here: pmi, t-test or llr
test = "pmi"
#here we select only those collocations that occur at least 4 times in the data
collocations = collocations[collocations.w12 >= 4]
print(collocations.shape)
#here we print out the top 30 (the number is set inside the head(n) function) collocations
print(collocations.sort_values(test, ascending=False).head(30))
```
