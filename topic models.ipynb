```python
import wget
import json, os, re, sys
import pandas, numpy
import urllib.parse
from topicmodels import *
```

This notebook browses LDA topic models generated for two datasets: Suomi24 internet discussion forum data ("suomi24" when used as a name in the code) and GlowBe online blog post data ("glowbe").

For both datasets, models have been generated for 5, 7, 10, 13, 15, 18, 20, 25 30 and 50 topics.

First, the topic model in question downloaded from github by running:

**download_topic_model(corpus, n_topics)**

and storing the result as the topic_model variable which is then used for further fuctions down the road. Corpus is either "glowbe" or "suomi24". Number of topics is given as a number and must be one of 5, 7, 10, 13, 15, 18, 20, 25 30 or 50.

The notebook then allows the following functions for browsing the model:

**1. view_topic_docs(topic_model, n_topics, [which_topics])**

lists the n_topics most heavy documents for topics. n_topics is given as a number and the topic for which documents are listed (which_topics) can defined as either a number (i.e. 5 for the topic number 5 etc), or as a list of numbers (i.e. [5,8,2] produces document lists for topics number 5, 8 and 2). If no topics are specified, all are printed.

The texts are identified by their id-identifiers, which are used when querying the texts for close reading.

**2. read_text(corpus, text_id)**

prints the text specified by identifier. Identifiers are something like "635050" for glowbe-corpus and "S24_2010_979108" for suomi24. Corpus is again identified as either "suomi24" or "glowbe".

**3.topics_over_word(topic_model, lemma)**

shows how a lemma distributes over topics.

**4.view_topic_words(topic_model, n_topics, [which_topics])**

same as 1. but for words instead of documents. Shows two columns: leftmost are the words which are absolutely most frequent in the topic, rightmost those which has proportionally biggest section in the topic. Same arguments specifications as 1.

```python
corpus = "suomi24"
n_topics = 10
topic_model = download_topic_model(corpus, n_topics)
```

```python
view_topic_docs(topic_model, 10, [3,5,8,9])
```

```python
read_text(corpus, "S24_2011_8764426")
```

```python
lemma = "nainen"
topics_over_word(topic_model, lemma)
```

```python
view_topic_words(topic_model, 20)
```
