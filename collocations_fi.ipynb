```python
import json, os, re, sys, codecs, time
import matplotlib.pyplot as plt
import collections
import pandas, numpy
from scipy.stats import binom
import wget
from korp import *
#from eng_corpus import *
```

```python
#the corpus is selected here, options are "suomi24" and "klk"
corpus = "suomi24"
crop = ""
```

```python
#necessary resources are built here, might take some time (around two minutes or so)
frequencies, tf = build_frequency_table(corpus)
```

```python
#the studied word is selected here, as well as the window size in words on either side (rang). 
#Also cropping is possible (ie. concentrating on particular subsection of the data based on metadata categories, 
#see data documentation document for which categories are available for which dataset)
lemma = "erehtyä"
rang = 5
```

```python
#the collocations are built here, can take some time (few minutes)
collocations = build_collocation_table(frequencies, tf, lemma, corpus, rang, crop)
```

```python
#only collocations with number of occurrences of at least 4 are selected
collocations = collocations[collocations.w12 >= 4]

#we can also introduce a high-frequency cap to do away with too common (and thus perhaps less informative)
#words like punctuation, pronouns, conjunctions etc
collocations = collocations[collocations.w2 < 10000000]
```

```python
#and listed here. The collocations are computed using three tests in parallel (pmi, t-test and llr). 
#and can be listed by any of them. head(n) function sets how many lines are printed.
test = "llr"
print(collocations.sort_values(test, ascending=False).head(30)[['llr', 'w12']])
```
